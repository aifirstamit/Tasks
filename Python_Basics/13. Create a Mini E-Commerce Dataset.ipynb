**Explain the combination of**

List

Dictionary

Indexing

Keys

Values

### Understanding the Combination


#### 1. List

The outer structure is a list.

products = [ ... ]

A list allows us to store multiple products.



#### 2. Dictionary

Each product is a dictionary.

{
    "name": "Laptop",
    "price": 70000,
    "brand": "Dell"
}

A dictionary stores information using key-value pairs.



#### 3. Indexing

The list uses indexes to access products.

products[0]
products[1]
products[2]

Remember:

Index 0 → First product
Index 1 → Second product
Index 2 → Third product

#### 4. Keys

Dictionary keys identify the type of information.

Examples:

"name"
"price"
"brand"
"rating"

 #### 5. Values

Values are the actual information stored under the keys.

Example:

"name": "Laptop"
"price": 70000
"brand": "Dell"

Here:

"name" → key
"Laptop" → value
"price" → key
70000 → value
"brand" → key
"Dell" → value
