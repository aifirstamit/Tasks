# Explanation of the Set Operations

## `union()`

Combines two sets and returns **all unique values**.

```python
python_students.union(java_students)
```

 **Trick to remember:**

> "Give me everyone from both groups."

---

## `intersection()`

Returns values that exist in **both sets**.

```python
python_students.intersection(java_students)
```

 **Trick to remember:**

> "Who is common to both groups?"

---

## `difference()`

Returns values that exist in the **first set but not the second set**.

```python
python_students.difference(java_students)
```

 **Think:**

> "Who is only in Python?"

###  The direction matters

These two operations produce different results:

```python
python_students.difference(java_students)
```

and

```python
java_students.difference(python_students)
```

The first means:

> Students in Python but **not** Java.

The second means:

> Students in Java but **not** Python.

---

## `symmetric_difference()`

Returns values that belong to **exactly one set**.

Students who are in **both sets are excluded**.

```python
python_students.symmetric_difference(java_students)
```

 **Trick to remember:**

> "Who is in one group, but not both?"

---

## `add()`

Adds **one item** to a set.

```python
python_students.add("Amit")
```

After adding:

```text
Before:
{"Rahul", "Aman", "Priya"}

After:
{"Rahul", "Aman", "Priya", "Amit"}
```

---

## `remove()`

Removes an item from a set.

```python
python_students.remove("Amit")
```

If the item **does not exist**, Python raises a `KeyError`.

```python
python_students.remove("John")
```

This will produce an error if `"John"` is not present in the set.

---

## `discard()`

Removes an item if it exists.

```python
python_students.discard("Abhisekh")
```

If the item **does not exist**, Python does **not** raise an error.

```python
python_students.discard("John")
```

This is safe even when `"John"` is not present.

---

##  Quick Memory Trick

| Method                   | Meaning                              |
| ------------------------ | ------------------------------------ |
| `union()`                | Everyone from both groups. → Everyone            |
| `intersection()`         | Common to both groups. → Both                |
| `difference()`           | Only in the first group. → Only first              |
| `symmetric_difference()` | In one group, but not both. → Only one           |
| `add()`                  | Add an item. → Add                          |
| `remove()`               | Remove an item — error if missing. → Remove strictly    |
| `discard()`              | Remove an item — no error if missing. → Remove safely |
