## Dictionary Methods Explained

### `keys()`

Returns all the keys in the dictionary.

```python
student.keys()
```

 **Think:**

> "What information categories do I have?"

---

### `values()`

Returns all the values in the dictionary.

```python
student.values()
```

 **Think:**

> "What information is stored?"

---

### `items()`

Returns all key-value pairs.

```python
student.items()
```

 **Think:**

> "Give me both the keys and their values."

---

### `get()`

Retrieves the value associated with a key.

```python
student.get("name")
```

If the key doesn't exist, `get()` returns `None` by default instead of raising an error.

 **Think:**

> "Give me the value for this key safely."

---
### `update()`

Adds new key-value pairs or updates existing values.

```python
student.update({"marks": 95})
```

This changes the marks to `95`.

We can also add a new key:

```python
student.update({"grade": "A"})
```

 **Think:**

> "Update or add information."

---

### `pop()`

Removes a key-value pair.

```python
student.pop("city")
```

This removes the `"city"` key and its value.

 **Think:**

> "Remove this specific key."

---

### `copy()`

Creates a copy of the dictionary.

```python
student_copy = student.copy()
```

 **Think:**

> "Make another dictionary with the same data."

---

###  Quick Reference

| Method     | Purpose                 | Example                         |
| ---------- | ----------------------- | ------------------------------- |
| `keys()`   | Get all keys            | `student.keys()`                |
| `values()` | Get all values          | `student.values()`              |
| `items()`  | Get key-value pairs     | `student.items()`               |
| `get()`    | Safely retrieve a value | `student.get("name")`           |
| `update()` | Add or update data      | `student.update({"marks": 95})` |
| `pop()`    | Remove a key-value pair | `student.pop("city")`           |
| `copy()`   | Create a copy           | `student.copy()`                |
