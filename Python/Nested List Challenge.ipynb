#   How indexing works with nested lists

A nested list is a list that contains other lists.

```text
students
   |
   |---- 0 → ["Rahul", 21, "AI"]
   |
   |---- 1 → ["Priya", 22, "Data Science"]
   |
   |---- 2 → ["Aman", 20, "Machine Learning"]
   |
   |---- 3 → ["Neha", 23, "Deep Learning"]
```

##  Indexing Inside Each Nested List

Each nested list has its own indexes:

```text
["Rahul", 21, "AI"]
     0      1      2

["Priya", 22, "Data Science"]
     0      1          2

["Aman", 20, "Machine Learning"]
     0      1            2
```

### How does nested indexing work?

```text
First index  → Which student's record?
Second index → Which piece of information?
```

For example:

```python
students[1][1]
```

This works in two steps:

```text
students[1]     → Priya's record
students[1][1]  → Priya's age
```

Therefore:

```python
students[1][1]
```

returns:

```text
22
```

###  Easy way to remember

```text
students[row][column]
     ↓       ↓
  Student   Information
```

So, `students[1][1]` means:

> Select the student at index `1` → **Priya**

> Then select the information at index `1` → **Age**

> Result → **22**
