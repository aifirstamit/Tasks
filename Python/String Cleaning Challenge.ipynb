##  Explanation: Python String Methods

### 1. `strip()`
Removes spaces from the **beginning and end** of a string.

### 2. `lower()`
Converts all characters to **lowercase**.

### 3. `upper()`
Converts all characters to **uppercase**.

### 4. `title()`
Capitalizes the **first letter of every word**.

### 5. `replace()`
Replaces a specified piece of text with another text.

### 6. `find()`
Finds the **position (index)** of a specified piece of text.
