````python
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
````

````python
print(numbers)
````

````python
set_numbers = set(numbers)
print(set_numbers)
````

#Duplicates disappeared : 10, 20 ,30

````python
list_numbers = list(set_numbers)
print(list_numbers)
````

````python
print(numbers)
number_of_original_elements = len(numbers)
print(number_of_original_elements)
````

````python
print(set_numbers)
number_of_unique_elements = len(set_numbers)
print(number_of_unique_elements)
````

## Why Are Sets Useful?

Sets are very useful when working with data that may contain duplicates.

For example, imagine a school has this list of students:

```text
students = ["Amit", "Rahul", "Priya", "Abhi", "Abhi", "Amit", "Amit", "Priya", "Abhi", "Rahul", "Rahul"]

If we want to know which different students are present, we don't need the repeated names.

We can simply use: set(students)
This gives us only the unique names.
{'Rahul', 'Amit', 'Priya', 'Abhi'}
