### Data type conversion or Typecasting

```python
int("100") #string "100" is changed into integer 100
```

```python
int("500") #string ""500" is changed into integer 500
```

```python
float("45.67") #string "45.67" is changed into float 45.67
```

```python
float("89.45") #string "89.45" is changed into float 89.45
```

```python
str(500) #integer 500 is changed into string "500"
```

```python
str(1000) #integer 1000 is changed into string "1000"
```

```python
bool(1) #integer 1 is changed into boolean True
```

```python
bool(0) #integer 0 is changed into boolean False
```

```python
list((1, 2, 3)) #tuple (1, 2, 3) is changed into list [1, 2, 3]
```

```python
list((9,45,67.54)) #tuple (9,45,67.54) is changed into to list [9, 45, 67.54]
```

```python
tuple([1, 2, 3]) #list [1, 2, 3] is changed into tuple (1, 2, 3)
```

```python
tuple([99, 2.2, 7.93]) #list a[99, 2.2, 7.93] is changed into tuple (99, 2.2, 7.93)
```

```python
tuple({34,67,90}) #set {34,67,90} is changed into tuple (34, 67, 90)
```

```python
tuple({67, 24, 99, 43}) #set 67, 24, 99, 43} is changed into tuple (24, 99, 67, 43)
```

```python
set([1, 2, 2, 3]) #list [1, 2, 2, 3]) is changed into set {1, 2, 3}
```

```python
set([21,78,34]) #list [21,78,34] is changed into set {21, 34, 78}
```
