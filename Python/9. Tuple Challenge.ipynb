##  Why Are Tuples Called Immutable?

**Immutable** means:

> Once a tuple is created, its individual elements cannot be changed, added, or removed.

### Example

```python
technologies = ("Python", "Java", "C++")
```

We cannot directly change `"Java"` to `"Go"`:

```python
technologies[1] = "Go"
```

Python will give an error:

```text
TypeError: 'tuple' object does not support item assignment
```

###  Why does this happen?

A tuple is **immutable**, so we cannot directly modify its elements.

Tuples do not support methods such as:

* `append()`
* `remove()`
* `insert()`
* `clear()`

because these methods would modify the tuple.

---

##  But We Can Create a New Tuple

If we need to modify the data, we can follow these three steps:

1. Convert the tuple → **list**
2. Modify the list
3. Convert the list → **tuple**

### Example

```python
technologies = ("Python", "Java", "C++")

# Step 1: Convert tuple into a list
technologies_list = list(technologies)

# Step 2: Modify the list
technologies_list.append("Go")

# Step 3: Convert the list back into a tuple
technologies = tuple(technologies_list)

print(technologies)
```

Output:

```text
('Python', 'Java', 'C++', 'Go')
```

###  Important Point

We did **not** modify the original tuple.

Instead, we:

```text
Tuple
  ↓
Convert to List
  ↓
Modify the List
  ↓
Convert back to Tuple
  ↓
New Tuple
```

Therefore:

> **Tuples are immutable because their elements cannot be directly changed after the tuple is created.**
