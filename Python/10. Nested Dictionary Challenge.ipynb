###  How Nested Dictionaries Are Accessed

A **nested dictionary** is simply a dictionary inside another dictionary.

Our structure looks like this:

```text
employee
│
├── name → "Amit"
│
├── department → "Engineering"
│
├── skills
│   │
│   ├── language → "Python + JavaScript"
│   ├── database → "PostgreSQL"
│   ├── cloud → "AWS"
│   └── framework → "Django"
│
├── salary → 90000
│
└── experience → 3

The main dictionary is:

employee

The "skills" key contains another dictionary:

employee["skills"]

To access something inside the nested dictionary, we use another key:

employee["skills"]["language"]

Think of it like going through two doors:

employee
   ↓
skills
   ↓
language
   ↓
"Python + JavaScript"

So:

employee["skills"]["language"]

means:

Go inside employee → find skills → then find language.

 One-Level vs Two-Level Access
One-level dictionary
employee["name"]

Only one key is required because "name" is directly inside employee.

Two-level nested dictionary
employee["skills"]["cloud"]

Two keys are required because "cloud" is inside the "skills" dictionary.

employee
   ↓
skills
   ↓
cloud
 Easy Rule to Remember
Normal dictionary:

dictionary["key"]


Nested dictionary:

dictionary["outer_key"]["inner_key"]

For our example:

employee["name"]

→ Accesses the employee's name.

employee["skills"]["language"]

→ Accesses the programming language.

employee["skills"]["database"]

→ Accesses the database.

employee["skills"]["cloud"]

→ Accesses the cloud technology.
